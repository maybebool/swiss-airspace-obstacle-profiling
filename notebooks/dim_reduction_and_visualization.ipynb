# Dimensionality Reduction & Visualization

**Objective:** Verify and visualise the K-Means k=4 clustering through dimensionality reduction (PCA, t-SNE) and geographic overlay maps. Confirm that clusters correspond to operationally meaningful airspace zones.

**Input:** `data/processed/grid_clustered.parquet` (3,976 cells with cluster labels from Notebook 03)
**Output:** Visualisations and maps for the final report. No new parquet, this notebook is analysis-only.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import folium
```

## Load Clustered Grid

```python
grid = pd.read_parquet('../data/processed/grid_clustered.parquet')
print(f"Loaded: {len(grid)} grid cells")

feature_cols = ['obstacle_count', 'height_mean', 'height_max', 'height_std',
                'elevation_mean', 'pct_linear', 'pct_vertical', 'pct_diversity',
                'pct_lighted', 'has_airport']

X = grid[feature_cols].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Cluster labels and names for plotting
cluster_labels = grid['cluster_kmeans'].values
CLUSTER_NAMES = {
    0: 'Plateau Transmission',
    1: 'Airport & Urban',
    2: 'Alpine High-Rise',
    3: 'Urban Vertical'
}
grid['cluster_name'] = grid['cluster_kmeans'].map(CLUSTER_NAMES)
print(grid['cluster_name'].value_counts())
```

## PCA: Variance Decomposition

PCA reveals how much of the total variance in the 10 features can be explained by fewer components, and which features drive each component.

```python
sns.set_style('whitegrid')
pca_full = PCA(random_state=42)
pca_full.fit(X_scaled)

# Explained variance
cumvar = np.cumsum(pca_full.explained_variance_ratio_) * 100

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.bar(range(1, 11), pca_full.explained_variance_ratio_ * 100, alpha=0.6, color='steelblue')
ax1.plot(range(1, 11), cumvar, 'ro-')
ax1.set_xlabel('Principal Component')
ax1.set_ylabel('Variance Explained (%)')
ax1.set_title('PCA Variance Explained')
ax1.set_xticks(range(1, 11))

for i, (var, cum) in enumerate(zip(pca_full.explained_variance_ratio_ * 100, cumvar)):
    ax1.text(i + 1, cum + 1.5, f'{cum:.0f}%', ha='center', fontsize=9)

# How many components for 90%?
n_95 = np.argmax(cumvar >= 95) + 1
ax1.axhline(y=95, color='gray', linestyle='--', alpha=0.5, label=f'90% at PC{n_95}')
ax1.legend()

# Feature loadings for PC1 and PC2
loadings = pd.DataFrame(
    pca_full.components_[:2].T,
    columns=['PC1', 'PC2'],
    index=feature_cols
)
loadings.plot(kind='barh', ax=ax2, alpha=0.7)
ax2.set_title('Feature Loadings (PC1 & PC2)')
ax2.set_xlabel('Loading')
ax2.axvline(x=0, color='black', linewidth=0.5)

plt.tight_layout()
plt.show()

print(f"\nPC1 explains {pca_full.explained_variance_ratio_[0]*100:.1f}% of variance")
print(f"PC2 explains {pca_full.explained_variance_ratio_[1]*100:.1f}% of variance")
print(f"Components needed for 95% variance: {n_95}")
```

### PCA Projection: Clusters in PC1/PC2 Space

```python
pca_2d = PCA(n_components=2, random_state=42)
pca_result = pca_2d.fit_transform(X_scaled)

fig, ax = plt.subplots(figsize=(10, 8))

colors = ['#4C72B0', '#ad0e0e', '#55A868', '#a19d2f']
for i, name in CLUSTER_NAMES.items():
    mask = cluster_labels == i
    ax.scatter(pca_result[mask, 0], pca_result[mask, 1],
               c=colors[i], s=15, alpha=0.5, label=f'{name} (n={mask.sum()})')

ax.set_xlabel(f'PC1 ({pca_2d.explained_variance_ratio_[0]*100:.1f}%)')
ax.set_ylabel(f'PC2 ({pca_2d.explained_variance_ratio_[1]*100:.1f}%)')
ax.set_title('K-Means Clusters in PCA Space')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

### PCA Biplot: Feature Vectors

The biplot overlays the feature loading vectors onto the PCA scatter, showing which features drive the separation between clusters.

```python
fig, ax = plt.subplots(figsize=(12, 9))

# Scatter
for i, name in CLUSTER_NAMES.items():
    mask = cluster_labels == i
    ax.scatter(pca_result[mask, 0], pca_result[mask, 1],
               c=colors[i], s=10, alpha=0.3, label=f'{name}')

# Loading vectors
loadings_2d = pca_2d.components_.T
scale = 4

for j, feat in enumerate(feature_cols):
    ax.arrow(0, 0, loadings_2d[j, 0] * scale, loadings_2d[j, 1] * scale,
             head_width=0.08, head_length=0.05, fc='black', ec='black', alpha=0.7)
    ax.text(loadings_2d[j, 0] * scale * 1.15, loadings_2d[j, 1] * scale * 1.15,
            feat, fontsize=9, ha='center', va='center',
            bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8))

ax.set_xlabel(f'PC1 ({pca_2d.explained_variance_ratio_[0]*100:.1f}%)')
ax.set_ylabel(f'PC2 ({pca_2d.explained_variance_ratio_[1]*100:.1f}%)')
ax.set_title('PCA Biplot - Feature Directions')
ax.legend(fontsize=10, loc='upper left')
ax.axhline(y=0, color='gray', linewidth=0.3)
ax.axvline(x=0, color='gray', linewidth=0.3)
plt.tight_layout()
plt.show()
```

## t-SNE: Non-Linear Projection

t-SNE preserves local neighbourhood structure and often reveals cluster separations that PCA cannot capture in 2D. We test two perplexity values to check robustness.

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 8))

for ax, perp in zip(axes, [10, 30, 60]):
    tsne = TSNE(n_components=2, perplexity=perp, random_state=42, max_iter=1000)
    tsne_result = tsne.fit_transform(X_scaled)

    for i, name in CLUSTER_NAMES.items():
        mask = cluster_labels == i
        ax.scatter(tsne_result[mask, 0], tsne_result[mask, 1],
                   c=colors[i], s=10, alpha=0.5, label=f'{name}')

    ax.set_title(f't-SNE (perplexity={perp})')
    ax.set_xlabel('Component 1')
    ax.set_ylabel('Component 2')
    ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

The t-SNE projections confirm that the four K-Means clusters represent genuinely separable groups in
feature space. At perplexity 30, each cluster forms a distinct region with minimal overlap, which the
linear PCA projection (50.7% variance) could not fully capture. The separation remains stable across
perplexity values (10, 30, 60), indicating that the cluster structure is robust and not an artefact of
a specific parameter choice. Plateau Transmission cells form the largest contiguous mass, while Alpine
High-Rise and Urban Vertical produce tightly grouped islands consistent with their specialised profiles.

### PCA + t-SNE Hybrid

Applying t-SNE on the PCA-reduced data (keeping components that explain 90% variance) reduces noise and speeds up computation.

```python
pca_95 = PCA(n_components=n_95, random_state=42)
X_pca = pca_95.fit_transform(X_scaled)

tsne_hybrid = TSNE(n_components=2, perplexity=60, random_state=42, max_iter=1000)
tsne_hybrid_result = tsne_hybrid.fit_transform(X_pca)

fig, ax = plt.subplots(figsize=(10, 8))

for i, name in CLUSTER_NAMES.items():
    mask = cluster_labels == i
    ax.scatter(tsne_hybrid_result[mask, 0], tsne_hybrid_result[mask, 1],
               c=colors[i], s=15, alpha=0.5, label=f'{name} (n={mask.sum()})')

ax.set_xlabel('Component 1')
ax.set_ylabel('Component 2')
ax.set_title(f'PCA ({n_95} components) + t-SNE Hybrid')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

### t-SNE HDBSCAN Comparison

Same t-SNE projection coloured by HDBSCAN labels for visual comparison with K-Means.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# K-Means
for i, name in CLUSTER_NAMES.items():
    mask = cluster_labels == i
    ax1.scatter(tsne_hybrid_result[mask, 0], tsne_hybrid_result[mask, 1],
                c=colors[i], s=10, alpha=0.5, label=f'{name}')
ax1.set_title('PCA + t-SNE - K-Means (k=4)')
ax1.legend(fontsize=9)

# HDBSCAN
hdb_labels = grid['cluster_hdbscan'].values
noise_mask = hdb_labels == -1
ax2.scatter(tsne_hybrid_result[noise_mask, 0], tsne_hybrid_result[noise_mask, 1],
            c='lightgray', s=8, alpha=0.3, label='Noise')
clustered_mask = ~noise_mask
scatter = ax2.scatter(tsne_hybrid_result[clustered_mask, 0], tsne_hybrid_result[clustered_mask, 1],
                      c=hdb_labels[clustered_mask], cmap='tab10', s=10, alpha=0.5)
ax2.set_title('PCA + t-SNE - HDBSCAN (k=13)')
ax2.legend(fontsize=9)

for ax in [ax1, ax2]:
    ax.set_xlabel('Component 1')
    ax.set_ylabel('Component 2')

plt.tight_layout()
plt.show()
```

## Geographic Overlay: K-Means Clusters

The cluster assignments projected back onto the Swiss map, with one subplot per cluster to show the spatial distribution of each zone.

```python
fig, axes = plt.subplots(2, 2, figsize=(20, 16))

for ax, (i, name) in zip(axes.flatten(), CLUSTER_NAMES.items()):
    # Background: all cells in light gray
    ax.scatter(grid['longitude'], grid['latitude'], c='lightgray', s=5, alpha=0.2)

    # Highlight this cluster
    mask = grid['cluster_kmeans'] == i
    ax.scatter(grid.loc[mask, 'longitude'], grid.loc[mask, 'latitude'],
               c=colors[i], s=12, alpha=0.6)

    ax.set_title(f'Cluster {i}: {name} (n={mask.sum()}, {mask.sum()/len(grid)*100:.1f}%)',
                 fontsize=13)
    ax.set_xlabel('Longitude')
    ax.set_ylabel('Latitude')
    ax.set_aspect('equal')

plt.tight_layout()
plt.show()
```

## Interactive Map

Folium map with colour-coded hexagonal cells. Hover for cell details. Save as HTML for the report.

```python
from folium.plugins import Fullscreen, MiniMap, MousePosition


color_map = {0: '#4C72B0', 1: '#DD8452', 2: '#55A868', 3: '#C44E52'}

m = folium.Map(location=[46.8, 8.2], zoom_start=8, tiles=None)

folium.TileLayer('CartoDB positron', name='Light').add_to(m)
folium.TileLayer('OpenStreetMap', name='Street').add_to(m)
folium.TileLayer(
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}',
    attr='Esri', name='Satellite'
).add_to(m)
folium.TileLayer(
    tiles='https://{s}.tile.opentopomap.org/{z}/{x}/{y}.png',
    attr='OpenTopoMap', name='Terrain'
).add_to(m)

layers = {
    0: folium.FeatureGroup(name='🔵 Plateau Transmission'),
    1: folium.FeatureGroup(name='🟠 Alpine High-Rise'),
    2: folium.FeatureGroup(name='🟢 Airport & Urban'),
    3: folium.FeatureGroup(name='🔴 Urban Vertical'),
}

for _, row in grid.iterrows():
    cluster_id = int(row['cluster_kmeans'])
    name = CLUSTER_NAMES[cluster_id]

    popup_text = (
        f"<b>{name}</b><br>"
        f"Obstacles: {row['obstacle_count']:.0f}<br>"
        f"Height mean: {row['height_mean']:.0f}m<br>"
        f"Elevation: {row['elevation_mean']:.0f}m AMSL<br>"
        f"Linear: {row['pct_linear']:.0f}%<br>"
        f"Vertical: {row['pct_vertical']:.0f}%<br>"
        f"Lighted: {row['pct_lighted']:.0f}%"
    )

    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=3,
        color=color_map[cluster_id],
        fill=True,
        fill_color=color_map[cluster_id],
        fill_opacity=0.6,
        popup=folium.Popup(popup_text, max_width=250),
    ).add_to(layers[cluster_id])

# Add all layers to map
for layer in layers.values():
    layer.add_to(m)

folium.LayerControl(collapsed=False).add_to(m)
Fullscreen().add_to(m)
MiniMap(toggle_display=True).add_to(m)
MousePosition(
    position='bottomright',
    prefix='Coordinates:',
    num_digits=4
).add_to(m)

legend_html = """
<div style="position: fixed; bottom: 30px; left: 30px; z-index: 1000;
     background: white; padding: 15px; border-radius: 5px;
     border: 2px solid grey; font-size: 13px; line-height: 1.8;">
<b>Cluster Assignments</b><br>
<span style="color: #4C72B0;">&#9632;</span> Plateau Transmission<br>
<span style="color: #DD8452;">&#9632;</span> Alpine High-Rise<br>
<span style="color: #55A868;">&#9632;</span> Airport &amp; Urban<br>
<span style="color: #C44E52;">&#9632;</span> Urban Vertical
</div>
"""
m.get_root().html.add_child(folium.Element(legend_html))

info_html = """
<div style="position: fixed; top: 10px; left: 50%; transform: translateX(-50%); z-index: 1000;">
<details style="background: white; padding: 12px; border-radius: 5px;
         border: 2px solid grey; font-size: 12px; max-width: 350px;
         line-height: 1.6; cursor: pointer;">
<summary style="font-weight: bold; font-size: 13px;">ℹ️ About this Map</summary>
<br>
This map shows K-Means (k=4) cluster assignments for 3,976 hexagonal grid cells
(H3 resolution 7, ~5 km² per cell). Each dot represents one grid cell, coloured
by its assigned airspace risk zone. Click any dot for detailed feature values.
<br><br>
This is the point-based cluster view from Notebook 04. For the full hexagonal
risk zone map with clear airspace overlay, see risk_zone_map.html.
</details>
</div>
"""
m.get_root().html.add_child(folium.Element(info_html))

slider_html = """
<div style="position: fixed; bottom: 5px; left: 50%; transform: translateX(-50%); z-index: 1000;
     background: white; padding: 12px; border-radius: 5px;
     border: 2px solid grey; font-size: 13px; width: 200px;">
<b>Dot Opacity</b><br>
<input type="range" id="opacitySlider" min="0" max="100" value="60"
       style="width: 100%; cursor: pointer;"
       oninput="
         var val = this.value / 100;
         document.getElementById('opacityValue').innerText = this.value + '%';
         document.querySelectorAll('path').forEach(function(p) {
           p.style.fillOpacity = val;
         });
       ">
<span id="opacityValue">60%</span>
</div>
"""
m.get_root().html.add_child(folium.Element(slider_html))

m.save('../docs/cluster_map.html')
print(f"Enhanced cluster map saved")
m
```

## Cluster Feature Comparison

Normalised feature values per cluster to visualise what makes each zone distinct.

```python
profiles = grid.groupby('cluster_kmeans')[feature_cols].mean()

# Normalise to 0-1 for comparison
profiles_norm = (profiles - profiles.min()) / (profiles.max() - profiles.min())

fig, ax = plt.subplots(figsize=(14, 6))

x = np.arange(len(feature_cols))
width = 0.2

for i, name in CLUSTER_NAMES.items():
    ax.bar(x + i * width, profiles_norm.loc[i], width,
           label=name, color=colors[i], alpha=0.7)

ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(feature_cols, rotation=45, ha='right')
ax.set_ylabel('Normalised Value (0–1)')
ax.set_title('Cluster Feature Profiles (Normalised)')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

## Conclusion

PCA reveals that 8 components are needed to capture 95% of the variance across the 10 grid features,
with PC1 (27.3%) and PC2 (23.4%) providing a partial but informative 2D view. In PCA space, the four
K-Means clusters show visible separation, with Plateau Transmission forming a compact low-variance
group and Alpine High-Rise spreading along the height-driven PC1 axis. t-SNE projections at multiple
perplexity values (10, 30, 60) confirm that all four clusters form genuinely separable groups in
feature space, with the separation remaining stable across parameter choices. Geographic overlay maps
validate the operational coherence of the segmentation: Plateau Transmission covers the Swiss Plateau,
Airport & Urban clusters around known airfields, Alpine High-Rise traces the alpine arc, and Urban
Vertical marks industrial sites. The normalised feature profiles confirm that each cluster carries a
unique signature with no redundancy. K-Means k=4 is confirmed as the primary result for corridor
profiling in Notebook 05.
